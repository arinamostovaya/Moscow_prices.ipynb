```python
rental_df = df[
    ["year_month", "district", "okrug",
     "rental_price_per_sqm_monthly", "n_listings_rental"]
]

rental_df.head(10)
```

```python
print("Количество пропусков:")
print(df.isnull().sum())

print("\nКоличество дубликатов:")
print(df.duplicated().sum())
```

```python
rental_df = df[
    [
        "year_month",
        "district",
        "okrug",
        "rental_price_per_sqm_monthly",
        "n_listings_rental"
    ]
].copy()

rental_df["year_month"] = pd.to_datetime(
    rental_df["year_month"],
    errors="coerce"
)

rental_df["rental_price_per_sqm_monthly"] = pd.to_numeric(
    rental_df["rental_price_per_sqm_monthly"],
    errors="coerce"
)

rental_df["n_listings_rental"] = pd.to_numeric(
    rental_df["n_listings_rental"],
    errors="coerce"
)

rental_df = rental_df.dropna(
    subset=["year_month", "district",
            "rental_price_per_sqm_monthly"]
)

rental_df = rental_df[
    rental_df["rental_price_per_sqm_monthly"] > 0
]

print("Размер очищенной таблицы:", rental_df.shape)

display(rental_df.head())
```

```python
print("Начало периода:", rental_df["year_month"].min())
print("Конец периода:", rental_df["year_month"].max())

print(
    "Количество районов:",
    rental_df["district"].nunique()
)

print(
    "Количество месяцев:",
    rental_df["year_month"].nunique()
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
monthly_prices = (
    rental_df.groupby("year_month")
    ["rental_price_per_sqm_monthly"]
    .mean()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_prices.index,
    monthly_prices.values,
    marker="o"
)

plt.title("Динамика стоимости аренды 1 м² в Москве")
plt.xlabel("Месяц")
plt.ylabel("Стоимость аренды, руб./м² в месяц")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
latest_month = rental_df["year_month"].max()

latest_data = rental_df[
    rental_df["year_month"] == latest_month
]

district_prices = (
    latest_data.groupby("district")
    ["rental_price_per_sqm_monthly"]
    .mean()
    .sort_values(ascending=False)
)

print("Самые дорогие районы:")
display(district_prices.head(10))

print("Самые доступные районы:")
display(district_prices.tail(10))
```

```python
plt.figure(figsize=(12, 7))

district_prices.head(10).sort_values().plot(
    kind="barh"
)

plt.title("10 самых дорогих районов Москвы")
plt.xlabel("Стоимость аренды, руб./м² в месяц")
plt.ylabel("Район")

plt.tight_layout()
plt.show()
```

```python
okrug_prices = (
    latest_data.groupby("okrug")
    ["rental_price_per_sqm_monthly"]
    .mean()
    .sort_values(ascending=False)
)

display(okrug_prices)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=okrug_prices.values,
    y=okrug_prices.index
)

plt.title("Стоимость аренды по административным округам")
plt.xlabel("Стоимость аренды, руб./м² в месяц")
plt.ylabel("Округ")

plt.tight_layout()
plt.show()
```

```python
listings_by_district = (
    latest_data.groupby("district")
    ["n_listings_rental"]
    .sum()
    .sort_values(ascending=False)
)

display(listings_by_district.head(10))

plt.figure(figsize=(12, 6))

listings_by_district.head(10).sort_values().plot(
    kind="barh"
)

plt.title("10 районов с наибольшим количеством объявлений")
plt.xlabel("Количество объявлений")
plt.ylabel("Район")

plt.tight_layout()
plt.show()
```

```python
latest_data = latest_data.copy()

latest_data["estimated_rent_40m2"] = (
    latest_data["rental_price_per_sqm_monthly"] * 40
)

display(
    latest_data[
        ["district", "rental_price_per_sqm_monthly",
         "estimated_rent_40m2"]
    ].sort_values(
        "estimated_rent_40m2",
        ascending=False
    ).head(10)
)
```

```python
monthly_change = monthly_prices.pct_change() * 100

print("Изменение стоимости аренды по месяцам:")
display(monthly_change.tail(12))
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=latest_data,
    x="n_listings_rental",
    y="rental_price_per_sqm_monthly"
)

plt.title("Стоимость аренды и количество объявлений по районам")
plt.xlabel("Количество объявлений")
plt.ylabel("Стоимость аренды, руб./м² в месяц")

plt.tight_layout()
plt.show()
```

```python
rental_df.to_csv(
    "moscow_rental_cleaned.csv",
    index=False,
    encoding="utf-8-sig"
)

district_prices.to_csv(
    "moscow_district_ranking.csv",
    encoding="utf-8-sig"
)

monthly_prices.to_csv(
    "moscow_monthly_rental_prices.csv",
    encoding="utf-8-sig"
)

print("Файлы успешно сохранены!")
```
